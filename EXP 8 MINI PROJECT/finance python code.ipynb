```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/Financial_Transactions_Clean.csv")

# Create a copy for processing
finance = df.copy()

print("First 5 Records:")
display(finance.head())

print("\nDataset Shape:")
print(finance.shape)

print("\nColumn Names:")
print(finance.columns.tolist())

print("\nData Types:")
print(finance.dtypes)

print("\nMissing Values:")
print(finance.isnull().sum())

print("\nDuplicate Records:")
print(finance.duplicated().sum())
```

```python
# ==============================
# STEP 2: DATA CLEANING
# ==============================

# Create another copy for cleaning
bank = finance.copy()

# Standardize column names
bank.columns = (
    bank.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

# Remove duplicate records
before = bank.shape[0]

bank.drop_duplicates(inplace=True)

after = bank.shape[0]

print("Duplicates removed:", before - after)

# Display missing values
print("\nMissing Values After Cleaning:")
print(bank.isnull().sum())

# Display cleaned dataset information
print("\nCleaned Dataset Shape:")
print(bank.shape)

print("\nCleaned Column Names:")
print(bank.columns.tolist())

print("\nFirst 5 Cleaned Records:")
display(bank.head())
```

```python
# ==============================
# STEP 2: DATA CLEANING
# ==============================

# Create another copy for cleaning
bank = finance.copy()

# Standardize column names
bank.columns = (
    bank.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

# Remove duplicate records
before = bank.shape[0]

bank.drop_duplicates(inplace=True)

after = bank.shape[0]

print("Duplicates removed:", before - after)

# Display missing values
print("\nMissing Values After Cleaning:")
print(bank.isnull().sum())

# Display cleaned dataset information
print("\nCleaned Dataset Shape:")
print(bank.shape)

print("\nCleaned Column Names:")
print(bank.columns.tolist())

print("\nFirst 5 Cleaned Records:")
display(bank.head())
```

```python
# ==============================
# STEP 3: DATE FORMATTING
# ==============================

# Convert transaction_date to datetime
bank["transaction_date"] = pd.to_datetime(
    bank["transaction_date"],
    errors="coerce"
)

# Check for invalid dates
print("Invalid dates:", bank["transaction_date"].isna().sum())

# Display data types
print("\nData Types:")
print(bank.dtypes)

# Display date range
print("\nTransaction Date Range:")
print("From:", bank["transaction_date"].min())
print("To:", bank["transaction_date"].max())

# Display sample records
print("\nSample Records:")
display(bank.head())
```

```python
# ==============================
# STEP 3: DATE FORMATTING
# ==============================

# Convert transaction_date into proper datetime format
bank["transaction_date"] = pd.to_datetime(
    bank["transaction_date"],
    errors="coerce"
)

# Check for invalid dates
print("Invalid dates:", bank["transaction_date"].isna().sum())

# Display data types
print("\nData Types:")
print(bank.dtypes)

# Display date range
print("\nTransaction Date Range:")
print("From:", bank["transaction_date"].min())
print("To:", bank["transaction_date"].max())
```

```python
# ==============================
# STEP 4: CREATE ANALYTICAL COLUMNS
# ==============================

# Extract year and month from transaction date
bank["year"] = bank["transaction_date"].dt.year
bank["month"] = bank["transaction_date"].dt.month
bank["month_name"] = bank["transaction_date"].dt.month_name()

# Create year-month column
bank["year_month"] = bank["transaction_date"].dt.to_period("M").astype(str)

# Display the new columns
print("New Analytical Columns Created:")
print([
    "year",
    "month",
    "month_name",
    "year_month"
])

# Display sample data
display(
    bank[
        [
            "transaction_id",
            "transaction_date",
            "amount",
            "transaction_type",
            "transaction_status",
            "year",
            "month_name",
            "year_month"
        ]
    ].head(10)
)
```

```python
# ==========================================
# STEP 5: FINANCIAL TRANSACTION ANALYSIS
# ==========================================

# Total number of transactions
total_transactions = bank["transaction_id"].nunique()

# Total transaction amount
total_amount = bank["amount"].sum()

# Average transaction amount
average_amount = bank["amount"].mean()

# Total customers
total_customers = bank["customer_id"].nunique()

# Credit transactions
credit_transactions = (
    bank[bank["transaction_type"].str.lower() == "credit"]
    ["transaction_id"].nunique()
)

# Debit transactions
debit_transactions = (
    bank[bank["transaction_type"].str.lower() == "debit"]
    ["transaction_id"].nunique()
)

# Completed transactions
completed_transactions = (
    bank[bank["transaction_status"].str.lower() == "completed"]
    ["transaction_id"].nunique()
)

# Failed transactions
failed_transactions = (
    bank[bank["transaction_status"].str.lower() == "failed"]
    ["transaction_id"].nunique()
)

# Display results
print("========== FINANCIAL ANALYTICS SUMMARY ==========")

print("Total Transactions       :", total_transactions)
print("Total Transaction Amount :", round(total_amount, 2))
print("Average Transaction      :", round(average_amount, 2))
print("Total Customers          :", total_customers)

print("\nTransaction Type:")
print("Credit Transactions      :", credit_transactions)
print("Debit Transactions       :", debit_transactions)

print("\nTransaction Status:")
print("Completed Transactions   :", completed_transactions)
print("Failed Transactions      :", failed_transactions)
```

```python
# ==========================================
# STEP 6: ACCOUNT TYPE ANALYSIS
# ==========================================

account_analysis = (
    bank.groupby("account_type")
    .agg(
        Total_Transactions=("transaction_id", "nunique"),
        Total_Amount=("amount", "sum"),
        Average_Amount=("amount", "mean")
    )
    .reset_index()
)

account_analysis["Total_Amount"] = account_analysis["Total_Amount"].round(2)
account_analysis["Average_Amount"] = account_analysis["Average_Amount"].round(2)

print("Account Type Analysis:")
display(account_analysis)
```

```python
# ==========================================
# STEP 7: MERCHANT ANALYSIS
# ==========================================

merchant_analysis = (
    bank.groupby("merchant")
    .agg(
        Total_Transactions=("transaction_id", "nunique"),
        Total_Amount=("amount", "sum"),
        Average_Amount=("amount", "mean")
    )
    .reset_index()
    .sort_values("Total_Amount", ascending=False)
)

merchant_analysis["Total_Amount"] = merchant_analysis["Total_Amount"].round(2)
merchant_analysis["Average_Amount"] = merchant_analysis["Average_Amount"].round(2)

print("Top Merchants by Transaction Amount:")
display(merchant_analysis)
```

```python
# ==========================================
# STEP 8: STANDARDIZE CATEGORICAL VALUES
# ==========================================

# Columns containing text/category values
categorical_columns = [
    "account_type",
    "transaction_type",
    "currency",
    "merchant",
    "transaction_status"
]

# Standardize text values
for col in categorical_columns:
    bank[col] = (
        bank[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .str.title()
    )

# Check account types after standardization
print("Account Types:")
print(bank["account_type"].value_counts())

print("\nTransaction Types:")
print(bank["transaction_type"].value_counts())

print("\nTransaction Status:")
print(bank["transaction_status"].value_counts())

print("\nMerchants:")
print(bank["merchant"].value_counts())
```

```python
# ==========================================
# STEP 9: UPDATED ACCOUNT ANALYSIS
# ==========================================

account_analysis = (
    bank.groupby("account_type")
    .agg(
        Total_Transactions=("transaction_id", "nunique"),
        Total_Amount=("amount", "sum"),
        Average_Amount=("amount", "mean")
    )
    .reset_index()
)

account_analysis["Total_Amount"] = account_analysis["Total_Amount"].round(2)
account_analysis["Average_Amount"] = account_analysis["Average_Amount"].round(2)

print("Updated Account Type Analysis:")
display(account_analysis)
```

```python
# ==========================================
# STEP 10: EXPORT CLEANED DATASET
# ==========================================

output_file = "/content/Financial_Transactions_Clean.csv"

bank.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_file)

print("\nFinal Dataset Shape:")
print(bank.shape)
```

```python
# ==========================================
# STEP 12: DATA VISUALIZATION
# ==========================================

import matplotlib.pyplot as plt

# Transaction Type Chart
transaction_type = bank.groupby("transaction_type")["transaction_id"].nunique()

transaction_type.plot(kind="bar")
plt.title("Credit vs Debit Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.show()
```

```python
# ==========================================
# STEP 13: MERCHANT TRANSACTION AMOUNT
# ==========================================

merchant_amount = bank.groupby("merchant")["amount"].sum().sort_values(ascending=False)

merchant_amount.plot(kind="bar")
plt.title("Transaction Amount by Merchant")
plt.xlabel("Merchant")
plt.ylabel("Total Transaction Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
# ==========================================
# STEP 14: ACCOUNT TYPE ANALYSIS CHART
# ==========================================

account_amount = bank.groupby("account_type")["amount"].sum()

account_amount.plot(kind="pie", autopct="%1.1f%%")
plt.title("Transaction Amount by Account Type")
plt.ylabel("")
plt.show()
```

```python
# ==========================================
# STEP 15: MONTHLY TRANSACTION TREND
# ==========================================

monthly_amount = bank.groupby("year_month")["amount"].sum()

monthly_amount.plot(kind="line", marker="o")
plt.title("Monthly Transaction Amount Trend")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")
plt.xticks(rotation=45)
plt.show()
```
