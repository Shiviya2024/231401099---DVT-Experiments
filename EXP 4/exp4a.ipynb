```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("city_day.csv")

# Display dataset
print("Dataset:")
print(df)

# Display first five records
print("\nFirst Five Records:")
print(df.head())

# Display shape
print("\nShape of Dataset:")
print(df.shape)

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Filter records with PM2.5 greater than 100
print("\nRecords with PM2.5 greater than 100:")
print(df[df["PM2.5"] > 100])

# Filter records with PM10 greater than 100
print("\nRecords with PM10 greater than 100:")
print(df[df["PM10"] > 100])

# Sort records by PM2.5
print("\nRecords Sorted by PM2.5:")
print(df.sort_values("PM2.5", ascending=False).head(10))

# Sort records by PM10
print("\nRecords Sorted by PM10:")
print(df.sort_values("PM10", ascending=False).head(10))

# City-wise average PM2.5
print("\nCity-wise Average PM2.5:")
print(df.groupby("City")["PM2.5"].mean())

# City-wise average PM10
print("\nCity-wise Average PM10:")
print(df.groupby("City")["PM10"].mean())

# Calculate mean
print("\nMean PM2.5:", df["PM2.5"].mean())

# Calculate median
print("Median PM2.5:", df["PM2.5"].median())

# Calculate minimum
print("Minimum PM2.5:", df["PM2.5"].min())

# Calculate maximum
print("Maximum PM2.5:", df["PM2.5"].max())

# Calculate standard deviation
print("Standard Deviation of PM2.5:", df["PM2.5"].std())

# Display correlation matrix
print("\nCorrelation Matrix:")
print(df[["PM2.5", "PM10"]].corr())

# Display correlation between PM2.5 and PM10
print("\nCorrelation between PM2.5 and PM10:")
print(df["PM2.5"].corr(df["PM10"]))

# Bar chart - PM2.5 for first 10 records
plt.figure(figsize=(10, 5))
plt.bar(df["City"].head(10), df["PM2.5"].head(10))
plt.title("PM2.5 Levels")
plt.xlabel("City")
plt.ylabel("PM2.5")
plt.xticks(rotation=45)
plt.show()

# Line chart - PM10 for first 10 records
plt.figure(figsize=(10, 5))
plt.plot(df["City"].head(10), df["PM10"].head(10), marker="o")
plt.title("PM10 Levels")
plt.xlabel("City")
plt.ylabel("PM10")
plt.xticks(rotation=45)
plt.show()

# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["PM2.5"], df["PM10"])
plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")
plt.show()

# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["PM2.5"].dropna(), bins=10)
plt.title("PM2.5 Distribution")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")
plt.show()

# Box plot
plt.figure(figsize=(6, 5))
plt.boxplot(df["PM2.5"].dropna())
plt.title("PM2.5 Box Plot")
plt.ylabel("PM2.5")
plt.show()
```
