```python
import pandas as pd
import matplotlib.pyplot as plt

# Import Dataset
df = pd.read_csv("/content/city_day.csv")

print("===== FIRST FIVE RECORDS =====")
display(df.head())


# =========================================================
# PROGRAM 1: LINE CHART
# =========================================================

plt.figure(figsize=(10,5))

plt.plot(
    df["PM2.5"].head(30),
    marker='o'
)

plt.title("PM2.5 Pollution Trend")
plt.xlabel("Record Number")
plt.ylabel("PM2.5")

plt.grid(True)

plt.show()


# =========================================================
# PROGRAM 2: BAR CHART
# =========================================================

city_pm25 = df.groupby("City")["PM2.5"].mean()

plt.figure(figsize=(10,5))

plt.bar(
    city_pm25.index,
    city_pm25.values
)

plt.title("Average PM2.5 by City")
plt.xlabel("City")
plt.ylabel("Average PM2.5")

plt.xticks(rotation=45)

plt.show()


# =========================================================
# PROGRAM 3: PIE CHART
# =========================================================

city_count = df["City"].value_counts()

plt.figure(figsize=(8,6))

plt.pie(
    city_count.values,
    labels=city_count.index,
    autopct='%1.1f%%'
)

plt.title("Air Quality Records by City")

plt.show()


# =========================================================
# PROGRAM 4: HISTOGRAM
# =========================================================

plt.figure(figsize=(8,5))

plt.hist(
    df["PM2.5"].dropna(),
    bins=30
)

plt.title("Distribution of PM2.5 Levels")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")

plt.show()


# =========================================================
# PROGRAM 5: SCATTER PLOT
# =========================================================

plot_data = df[["PM2.5", "PM10"]].dropna()

plt.figure(figsize=(8,5))

plt.scatter(
    plot_data["PM2.5"],
    plot_data["PM10"],
    alpha=0.3
)

plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")

plt.show()


# =========================================================
# PROGRAM 6: BOX PLOT
# =========================================================

plt.figure(figsize=(7,5))

plt.boxplot(
    df["PM2.5"].dropna()
)

plt.title("PM2.5 Box Plot")
plt.ylabel("PM2.5")

plt.show()


# =========================================================
# PROGRAM 7: MULTIPLE LINE CHART
# =========================================================

data = df[["PM2.5", "PM10"]].head(30)

plt.figure(figsize=(10,5))

plt.plot(
    data["PM2.5"].values,
    marker='o',
    label="PM2.5"
)

plt.plot(
    data["PM10"].values,
    marker='o',
    label="PM10"
)

plt.title("PM2.5 and PM10 Pollution Levels")
plt.xlabel("Record Number")
plt.ylabel("Pollution Level")

plt.legend()

plt.grid(True)

plt.show()
```
