```python
# ============================================================
# EXPERIMENT 2 - EDA: DATA IMPORT AND EXPORT
# CITY DAY AIR QUALITY DATASET
# ============================================================

import pandas as pd
import sqlite3

# Load the City Day dataset
df = pd.read_csv("city_day.csv")

# Display complete dataset
print("===== DATASET =====")
print(df)

# Display first five records
print("\n===== FIRST FIVE ROWS =====")
print(df.head())

# Display last five records
print("\n===== LAST FIVE ROWS =====")
print(df.tail())

# Display dataset information
print("\n===== DATASET INFORMATION =====")
df.info()

# Display statistical summary
print("\n===== STATISTICAL SUMMARY =====")
print(df.describe(include='all'))

# Display dataset shape
print("\nDataset Shape:", df.shape)

# Display column names
print("\n===== COLUMN NAMES =====")
print(df.columns.tolist())

# Check missing values
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Export dataset to CSV
df.to_csv("city_day_export.csv", index=False)
print("\nCSV file exported successfully!")

# Export dataset to Excel
df.to_excel("city_day_export.xlsx", index=False)
print("Excel file exported successfully!")

# Create SQLite database
conn = sqlite3.connect("city_day.db")

# Export DataFrame to SQL table
df.to_sql(
    "city_day",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset exported successfully to SQLite database!")

# Close database connection
conn.close()

print("SQLite database connection closed successfully!")
```
